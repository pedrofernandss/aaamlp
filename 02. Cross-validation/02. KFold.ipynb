K-Fold Cross Validation

Nós podemos separar o nosso conjunto de dados em dois conjuntos diferentes: treinamento e validação. Muita gente costuma também separar em um terceiro conjunto, o de teste. Entretanto, vamos trabalhar só com dois por enquanto.

Nós podemos dividir nosso dataset em K conjuntos diferentes de mesmo tamanho. Essa estratégia é conhecida como **K-Fold Cross Validation** e cada amostra é numerada de 0 à K-1

```python
import pandas as pd
from sklearn import model_selection
```

```python
df = pd.read_csv("winequality-red.csv")
```

```python
# Crio uma coluna KFold e preencho ela com o valor -1
df["kfold"] = -1
df = df.sample(frac=1).reset_index(drop=True)
```

```python
# Indico que serão 5 folds
kf = model_selection.KFold(n_splits=5)

# Indico em qual fold cada dado estará
for fold, (trn_, val_) in enumerate(kf.split(X=df)):
    df.loc[val_, 'kfold'] = fold
```

```python
df
```

Nessa técnica nós treinamos nosso modelo em K-1 folds e testamos no último. Esse processo é repetido K vezes e em cada uma delas um fold diferente é usado como conjunto de teste. Após isso, a performance do modelo é então a média da performance em cada uma das K iterações. Essa ideia tem como objetivo estimar de forma robusta a capacidade de generalização do modelo em relação aos dados.

Acredito que essa estratégia seja muito boa em datasets pequenos ou quando a aquisição dos dados é muito custosa pois cada dado é usado tanto para treinamento como para avaliação. Entretanto, é preciso tomar cuidado para o risco de data leakage, ou seja, que nenhum pre-processamento dos dados faça com que uma informação do conjunto de teste vaze para o conjunto de treinamento.
